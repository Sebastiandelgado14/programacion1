```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return -2*x**2 + 20*x

def f2(x):
    return -2*x**2 + 20*x + 10

x = np.linspace(0, 12, 400)
y = f(x)
y2 = f2(x)

plt.figure(figsize=(10, 6))
plt.plot(x, y, label='f(x) = -2x² + 20x', color='blue')
plt.plot(x, y2, label='f2(x) = -2x² + 20x + 10', color='red', linestyle='--')


plt.scatter([0, 5, 10], [0, 50, 0], color='blue', zorder=5)
plt.axvline(x=5, color='green', linestyle=':', label='Eje de simetría x=5')
plt.axhline(y=0, color='black', linewidth=0.5)

plt.title('Ganancia diaria vs Número de empanadas vendidas')
plt.xlabel('Número de empanadas (x)')
plt.ylabel('Ganancia (miles de pesos)')
plt.legend()
plt.grid(True)
plt.show()
```
