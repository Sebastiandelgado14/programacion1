```python
import re

# Excepción personalizada
class CorreoInvalidoError(Exception):
    pass

# Función para validar correo
def validar_correo(correo):

    patron = r"^[a-z]+_[0-9]+_[a-z]+@universitaria\.edu\.co$"

    if not re.fullmatch(patron, correo):
        raise CorreoInvalidoError(
            "Correo inválido. Debe tener el formato nombre_edad_apellido@universitaria.edu.co")

    return True

# Número de intentos
intentos = 3

# Programa principal
while intentos > 0:

    try:
        correo = input("Ingresar correo institucional: ")

        validar_correo(correo)

        print("Correo válido")
        break

    except CorreoInvalidoError as e:
        intentos -= 1
        print("Error:", e)

        if intentos > 0:
            print(f"Te quedan {intentos} intentos.")

    except Exception as e:
        print("Error inesperado:", e)

# Si falla los 3 intentos
if intentos == 0:
    print(
        "Acceso bloqueado por intentos fallidos")
```
